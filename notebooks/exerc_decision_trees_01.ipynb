# Data Import

Check data on crypto currencies in https://data.binance.vision/?prefix=data/spot/ .

```python
def get_pair_data(pair, start_date, end_date, time_interval = '1m'):
# Convert strings to datetime if needed
    if isinstance(start_date, str):
        start_date = dt.datetime.strptime(start_date, '%Y-%m-%d')
    if isinstance(end_date, str):
        end_date = dt.datetime.strptime(end_date, '%Y-%m-%d')
    
    # Generate list of dates
    date_list = []
    current_date = start_date
    
    while current_date <= end_date:
        date_list.append(current_date)
        current_date += dt.timedelta(days=1)

    kline_cols = [
            "open",
            "high",
            "low",
            "close",
            "volume",
            "timestamp_close",
            "quote_volume",
            "n_trades",
            "taker_buy_base_volume",
            "taker_buy_quote_volume",
            "ignore",
        ]

    kline_data = {}
    kline_data[pair] = pd.DataFrame(columns=kline_cols)
    kline_data[pair].index.name = 'timestamp'
    for date in tqdm(date_list):
        date_str = date.strftime("%Y-%m-%d")
        url = f"https://data.binance.vision/data/spot/daily/klines/{pair}/{time_interval}/{pair}-{time_interval}-{date_str}.zip"
        temp = pd.read_csv(url, index_col=0, header=None, names=kline_cols)
        temp.index = pd.to_datetime(temp.index * 1e-6, unit='s')
        for i, r in temp.iterrows():
            kline_data[pair].loc[r.name] = r
    return kline_data[pair]
```

```python
df = get_pair_data('BTCUSDC', '2025-01-01', '2025-12-31', '1h')
```

```python
df.to_csv('BTCUSDC_1h_2025.csv')
```
