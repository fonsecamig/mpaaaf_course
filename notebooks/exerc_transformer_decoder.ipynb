# Transformer Decoder

## 1.
Consider the dataset relative to daily pricing data on open, high, low, close, and volume for the S&P 500, NASDAQ Composite, Dow Jones Industrial Average, FTSE 100, Nikkei 225, Hang Seng Index, Swiss Market Index, DAX Index, CAC 40 and Euro Stoxx 50, from 2015-1-1 to 2025-12-31. Download the data and preprocess it by standardizing it into training and testing sets.

## 2.
Using a transformer decoder, train a model to predict index open, high, low, close, and volume values for the following day based on data from the previous 60 days. Report the training and validation loss curves, and evaluate the model on the test set using the RMSE and MAE metrics for the unstandardized values. Tune the model hyperparameters to optimize performance. Discuss the results and any challenges faced during training.

## 3.
Using the trained transformer decoder model, generate predictions for the next 30 days of index open, high, low, close, and volume values. Plot the predicted values against the actual values for the test set and analyze the performance of the model in capturing trends and patterns in the data. Discuss any discrepancies observed and potential reasons for them.
